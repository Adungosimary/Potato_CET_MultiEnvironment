## Reading every Excel file

```python
import pandas as pd

all_data = {}

for trial, info in trials.items():

    path = RAW / info["file"]

    xls = pd.ExcelFile(path)

    all_data[trial] = {}

    for sheet in xls.sheet_names:

        df = pd.read_excel(path, sheet_name=sheet)

        all_data[trial][sheet] = df

print("Finished reading all datasets.")
```
