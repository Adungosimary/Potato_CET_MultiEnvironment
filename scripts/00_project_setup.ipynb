## Building the project dictionary

```python
trials = {

"KE25MOL-EA21":
    {"file":"KE25MOL-EA21-AT01_table-Final.xlsx",
     "location":"Molo",
     "population":"EA21"},

"KE25MOL-LBHT":
    {"file":"KE25MOL-LBHT-IT01_table-Final.xlsx",
     "location":"Molo",
     "population":"LBHT"},

"KE25NJO-EA21":
    {"file":"KE25NJO-EA21-AT01_table-Final.xlsx",
     "location":"Njoro",
     "population":"EA21"},

"KE25NJO-LBHT":
    {"file":"KE25NJO-LBHT-IT01_table-Final.xlsx",
     "location":"Njoro",
     "population":"LBHT"},

"KE25UON-EA21":
    {"file":"KE25UON-EA21-AT01_table-Final.xlsx",
     "location":"Nairobi",
     "population":"EA21"},

"KE25UON-LBHT":
    {"file":"KE25UON-LBHT-IT01_table-Final.xlsx",
     "location":"Nairobi",
     "population":"LBHT"},
}
```

## Reading every Excel file

```python
import pandas as pd

all_data = {}

for trial, info in trials.items():

    path = RAW / info["file"]

    xls = pd.ExcelFile(path)

    all_data[trial] = {}

    for sheet in xls.sheet_names:

        df = pd.read_excel(path, sheet_name=sheet)

        all_data[trial][sheet] = df

print("Finished reading all datasets.")
```

Checking what was imported

```python
for trial in all_data:

    print("="*60)

    print(trial)

    for sheet in all_data[trial]:

        print(sheet,
              all_data[trial][sheet].shape)
```

## Saving backup CSVs

```python
for trial in all_data:

    for sheet, df in all_data[trial].items():

        name = f"{trial}_{sheet}.csv"

        # make filename safe
        name = (
            name.replace(" ", "_")
                .replace("/", "_")
                .replace("-", "_")
        )

        df.to_csv(CLEAN / name, index=False)

print("CSV backup complete.")
```
