## Reading every Excel file automatically

```python
trial_data = {}

for file in files:

    trial = Path(file).stem
    trial_data[trial] = {}

    print(f"\nProcessing: {trial}")

    xls = pd.ExcelFile(file)

    for sheet in xls.sheet_names:

        df = pd.read_excel(file, sheet_name=sheet)

        # Skip empty sheets
        if df.empty or df.dropna(how="all").empty:
            print(f"   Skipping empty sheet: {sheet}")
            continue

        if df.shape[1] >= 70:
            trial_data[trial]["yield"] = df
            print(f"   Yield      -> {sheet} ({df.shape[0]} × {df.shape[1]})")
        else:
            trial_data[trial]["phenotype"] = df
            print(f"   Phenotype -> {sheet} ({df.shape[0]} × {df.shape[1]})")
```

## Standardizing column names

```python
#==========================================================
# CELL 13
# Standardize column names
#==========================================================

import re

def clean_names(df):

    df = df.copy()

    df.columns = (
        df.columns
          .str.strip()
          .str.lower()
          .str.replace(r"[^\w]+", "_", regex=True)
          .str.replace(r"_+", "_", regex=True)
          .str.strip("_")
    )

    return df
```

## Cleaning every sheet

```python
#==========================================================
# CELL 14
# Clean every worksheet
#==========================================================

clean_trials = {}

for trial_name, sheets_dict in datasets.items():

    clean_trials[trial_name] = {}

    for sheet_name, df in sheets_dict.items():

        # Skip empty worksheets
        if df.empty:
            continue

        df = clean_names(df)

        # Remove empty rows
        df = df.dropna(how="all")

        # Remove empty columns
        df = df.dropna(axis=1, how="all")

        clean_trials[trial_name][sheet_name] = df

print("Cleaning complete.")
```

## Function to identify phenotype and yield sheets
instead of asuming the first sheet is phenotype, let's detect them automatically

```python
#==========================================================
# CELL 17
# Identify phenotype and yield sheets
#==========================================================

def identify_sheets(sheet_dict):

    phenotype = None
    yield_sheet = None

    for name in sheet_dict.keys():

        if "yield" in name.lower():

            yield_sheet = name

        else:

            phenotype = name

    return phenotype, yield_sheet
```

## Merging the datasets

```python
#==========================================================
# CELL 18
# Merge phenotype and yield
#==========================================================

merge_keys = [
    "plot",
    "geno",
    "entry",
    "row",
    "column",
    "rep"
]

merged_trials = {}

for trial_name, sheets_dict in clean_trials.items():

    pheno_sheet_name, yield_sheet_name = identify_sheets(sheets_dict)

    pheno_df = None
    yield_df = None

    if pheno_sheet_name:
        pheno_df = sheets_dict[pheno_sheet_name]
    if yield_sheet_name:
        yield_df = sheets_dict[yield_sheet_name]

    # Handle various combinations of sheet presence
    if pheno_df is None and yield_df is None:
        print(f"{trial_name} --> No phenotype or yield sheets found. Skipping.")
        continue
    elif pheno_df is None: # Only yield data available
        merged_trials[trial_name] = yield_df.copy()
        print(f"{trial_name} --> yield only")
        continue
    elif yield_df is None: # Only phenotype data available
        merged_trials[trial_name] = pheno_df.copy()
        print(f"{trial_name} --> phenotype only")
        continue
    else: # Both phenotype and yield data available, perform merge
        merged = pheno_df.merge(
            yield_df,
            on=merge_keys,
            how="left",
            suffixes=("", "_yield")
        )

        merged_trials[trial_name] = merged

        print(
            f"{trial_name}: "
            f"{pheno_df.shape[0]} -> {merged.shape[0]} rows"
        )
```

## Automatically converting numeric variables

```python
for trial in clean_trials:

    for sheet in clean_trials[trial]:

        df = clean_trials[trial][sheet]

        # Iterate using positional index to handle duplicate column names
        for i, col_name in enumerate(df.columns):
            # Access the specific column by its positional index
            col_series = df.iloc[:, i]

            # Check the dtype of this specific column (which is a Series)
            if col_series.dtype == object:

                converted = pd.to_numeric(col_series, errors="coerce")

                # Convert only if at least 70% of non-missing values are numeric
                non_missing = col_series.notna().sum()

                if non_missing > 0:
                    if converted.notna().sum() / non_missing >= 0.70:
                        df.iloc[:, i] = converted  # Assign back by positional index

        clean_trials[trial][sheet] = df

print("✅ Numeric conversion completed.")
```

## Add trial information

```python
master = []

# Helper function to ensure unique column names for a DataFrame
def make_columns_unique(df):
    cols = df.columns
    seen_counts = {}
    new_cols = []
    for col in cols:
        if col in seen_counts:
            seen_counts[col] += 1
            new_cols.append(f"{col}_{seen_counts[col]}")
        else:
            seen_counts[col] = 0 # First occurrence, count is 0 for no suffix
            new_cols.append(col)
    df.columns = new_cols
    return df

for trial_name, trial_df in merged_trials.items():
    # Make a copy to avoid modifying the original DataFrame in merged_trials
    current_df = trial_df.copy()

    # Ensure column names are unique within this DataFrame
    current_df = make_columns_unique(current_df)

    current_df["trial"] = trial_name

    location_prefix = trial_name.split("-")[0]

    if "MOL" in location_prefix:
        current_df["location"] = "Molo"
    elif "NJO" in location_prefix:
        current_df["location"] = "Njoro"
    elif "UON" in location_prefix:
        current_df["location"] = "Nairobi"

    # Extract population (e.g., EA21 from KE25MOL-EA21-AT01_table-Final)
    genotype_set = trial_name.split("-")[1]
    current_df["population"] = genotype_set

    master.append(current_df)

master_df = pd.concat(master, ignore_index=True)

print(master_df.shape)
```
